```python
%sql
CREATE OR REPLACE TABLE general_catalog.gold.mv_sales_by_territory AS
SELECT 
    t.TerritoryID,
    t.Name,
    SUM(f.TotalDue) AS total_revenue
FROM general_catalog.gold.fact_sales f
JOIN general_catalog.gold.dim_territory t ON f.TerritoryID = t.TerritoryID
GROUP BY t.TerritoryID, t.Name;
```
