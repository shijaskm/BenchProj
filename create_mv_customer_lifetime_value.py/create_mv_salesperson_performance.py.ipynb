```python
%sql
CREATE OR REPLACE TABLE general_catalog.gold.mv_salesperson_performance AS
SELECT 
    sp.BusinessEntityID,
    sp.FirstName,
    sp.LastName,
    SUM(f.TotalDue) AS total_sales,
    COUNT(DISTINCT f.SalesOrderID) AS total_orders
FROM general_catalog.gold.fact_sales f
JOIN general_catalog.gold.dim_salesperson sp ON f.SalesPersonID = sp.BusinessEntityID
GROUP BY sp.BusinessEntityID, sp.FirstName, sp.LastName;
```
