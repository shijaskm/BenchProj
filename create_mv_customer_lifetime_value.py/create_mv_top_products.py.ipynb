```python
%sql
CREATE OR REPLACE TABLE general_catalog.gold.mv_top_products AS
SELECT 
    p.ProductID,
    p.Name,
    SUM(f.LineTotal) AS total_sales
FROM general_catalog.gold.fact_sales f
JOIN general_catalog.gold.dim_product p ON f.ProductID = p.ProductID
GROUP BY p.ProductID, p.Name
ORDER BY total_sales DESC
LIMIT 10;
```
