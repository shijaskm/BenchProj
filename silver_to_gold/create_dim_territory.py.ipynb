```python
# Databricks notebook source

from pyspark.sql import functions as F
from pyspark.sql.window import Window
from pyspark.sql.functions import to_date, date_format


def create_gold_dim_Territory(Territory_tbl_df):

    Territory_tbl_df = Territory_tbl_df.select("TerritoryID", "Name", "CountryRegionCode", "Group", "SalesYTD", "SalesLastYear")
    Territory_tgt_df = Territory_tbl_df.dropDuplicates(["TerritoryID"]).withColumn("processed_timestamp", F.current_timestamp())
                                 
    return Territory_tgt_df



if __name__ == "__main__":

    Territory_tbl = dbutils.widgets.get("Territory")
    Territory_tbl_df = df = spark.read.table(Territory_tbl)
    Territory_tgt_df = create_gold_dim_Territory(Territory_tbl_df)
    Territory_gold_tbl = dbutils.widgets.get("Territory_gold_tbl")
    Territory_tgt_df.write.mode("overwrite").format("delta").partitionBy("TerritoryID").saveAsTable(Territory_gold_tbl)
```
