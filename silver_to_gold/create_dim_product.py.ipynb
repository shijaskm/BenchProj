```python
# Databricks notebook source

from pyspark.sql import functions as F
from pyspark.sql.window import Window
from pyspark.sql.functions import to_date, date_format


def create_gold_dim_product(Product_tbl_df):

    Product_tbl_df = Product_tbl_df.select("ProductID", "ProductNumber", "Name", "Color", "Size", "ProductSubcategoryID", "StandardCost", "ListPrice")
    product_tgt_df = Product_tbl_df.dropDuplicates(["ProductID"]).withColumn("processed_timestamp", F.current_timestamp())
                                 
    return product_tgt_df



if __name__ == "__main__":

    Product_tbl = dbutils.widgets.get("Product")
    Product_tbl_df = df = spark.read.table(Product_tbl)
    product_tgt_df = create_gold_dim_product(Product_tbl_df)
    product_gold_tbl = dbutils.widgets.get("Product_gold_tbl")
    product_tgt_df.write.mode("overwrite").format("delta").partitionBy("ProductID").saveAsTable(product_gold_tbl)
```
