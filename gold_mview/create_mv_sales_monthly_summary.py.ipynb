```python
%sql
CREATE OR REPLACE TABLE general_catalog.gold.mv_sales_monthly_summary AS
SELECT 
    substr(d.date_id,1,4) AS year,
    substr(d.date_id,5,2) AS month,
    d.date_id AS date_id,
    COUNT(DISTINCT f.SalesOrderID) AS total_orders,
    SUM(f.TotalDue) AS total_revenue
FROM general_catalog.gold.fact_sales f
JOIN general_catalog.gold.dim_date d ON f.OrderDate = d.OrderDate
GROUP BY year, month,date_id;
```
