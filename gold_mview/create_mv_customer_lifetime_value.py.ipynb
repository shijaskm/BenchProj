```python
%sql
CREATE OR REPLACE TABLE general_catalog.gold.mv_customer_lifetime_value AS
SELECT 
    c.CustomerId,
    c.FirstName,
    c.LastName,
    SUM(f.TotalDue) AS lifetime_value
FROM general_catalog.gold.fact_sales f
JOIN general_catalog.gold.dim_customer c ON f.CustomerId = c.CustomerId
GROUP BY c.CustomerId, c.FirstName, c.LastName;
```
